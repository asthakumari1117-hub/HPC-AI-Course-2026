```python
%%writefile memory_representation_of_string.c
#include <stdio.h>

int main()
{
    // ----------------------------------------------------
    // Program: Memory Representation of a String
    // Every character occupies one byte of memory.
    // The last byte stores the null character ('\0').
    // ----------------------------------------------------

    char city[] = "Delhi";

    printf("Index\tCharacter\tASCII\tMemory Address\n");
    printf("-----------------------------------------------\n");

    for(int i = 0; i <= 5; i++)
    {
        if(city[i] == '\0')
            printf("%d\t\\0\t\t%d\t%p\n", i, city[i], &city[i]);
        else
            printf("%d\t%c\t\t%d\t%p\n", i, city[i], city[i], &city[i]);
    }

    return 0;
}

!gcc memory_representation_of_string.c -o memory_representation_of_string
!./memory_representation_of_string
```

```python
%%writefile memory_layout_of_strings.c
#include <stdio.h>

int main()
{
    // ----------------------------------------------------
    // Program: Memory Layout of Strings
    // Display each character, its ASCII value,
    // and its memory address.
    // ----------------------------------------------------

    char word[] = "Computer";

    printf("Index\tCharacter\tASCII\tAddress\n");
    printf("-------------------------------------------------\n");

    for(int i = 0; word[i] != '\0'; i++)
    {
        printf("%d\t%c\t\t%d\t%p\n",
               i,
               word[i],
               word[i],
               &word[i]);
    }

    printf("%lu\t\\0\t\t0\t%p\n",
           sizeof(word)-1,
           &word[sizeof(word)-1]);

    return 0;
}

!gcc memory_layout_of_strings.c -o memory_layout_of_strings
!./memory_layout_of_strings
```

```python
%%writefile scanf_vs_fgets.c
#include <stdio.h>

int main()
{
    // ----------------------------------------------------
    // Program: Difference between scanf() and fgets()
    // scanf() reads input until the first whitespace.
    // fgets() reads an entire line including spaces.
    // ----------------------------------------------------

    char name1[50];
    char name2[50];

    printf("Enter your full name using scanf(): ");
    scanf("%49s", name1);

    getchar();   // Remove newline left by scanf()

    printf("Enter your full name using fgets(): ");
    fgets(name2, sizeof(name2), stdin);

    printf("\nOutput using scanf() : %s\n", name1);
    printf("Output using fgets() : %s", name2);

    return 0;
}

!gcc scanf_vs_fgets.c -o scanf_vs_fgets
!./scanf_vs_fgets
```

```python
%%writefile mini_project_text_analyzer.c
#include <stdio.h>
#include <string.h>
#include <ctype.h>

int main()
{
    // ----------------------------------------------------
    // Mini Project: Text Analyzer
    //
    // This program:
    // 1. Reads a sentence.
    // 2. Counts alphabets.
    // 3. Counts digits.
    // 4. Counts spaces.
    // 5. Counts uppercase letters.
    // 6. Counts lowercase letters.
    // 7. Displays the total length.
    // ----------------------------------------------------

    char text[200];

    int alphabets = 0;
    int digits = 0;
    int spaces = 0;
    int uppercase = 0;
    int lowercase = 0;

    printf("Enter a sentence:\n");
    fgets(text, sizeof(text), stdin);

    for(int i = 0; text[i] != '\0'; i++)
    {
        if(isalpha(text[i]))
            alphabets++;

        if(isdigit(text[i]))
            digits++;

        if(isspace(text[i]))
            spaces++;

        if(isupper(text[i]))
            uppercase++;

        if(islower(text[i]))
            lowercase++;
    }

    printf("\n========== Text Analysis ==========\n");

    printf("Input Text        : %s", text);
    printf("Total Length      : %lu\n", strlen(text));
    printf("Alphabets         : %d\n", alphabets);
    printf("Digits            : %d\n", digits);
    printf("White Spaces      : %d\n", spaces);
    printf("Uppercase Letters : %d\n", uppercase);
    printf("Lowercase Letters : %d\n", lowercase);

    return 0;
}

!gcc mini_project_text_analyzer.c -o mini_project_text_analyzer
!./mini_project_text_analyzer
```

```python
%%writefile sprintf_vs_snprintf.c
#include <stdio.h>

int main()
{
    // ----------------------------------------------------
    // Program: sprintf() vs snprintf()
    //
    // sprintf() does not check buffer size.
    // snprintf() limits the number of characters
    // written into the destination buffer.
    // ----------------------------------------------------

    char buffer1[100];
    char buffer2[20];

    sprintf(buffer1,
            "Welcome to C Programming Course");

    snprintf(buffer2,
             sizeof(buffer2),
             "Welcome to C Programming Course");

    printf("Using sprintf():\n");
    printf("%s\n\n", buffer1);

    printf("Using snprintf():\n");
    printf("%s\n", buffer2);

    return 0;
}

!gcc sprintf_vs_snprintf.c -o sprintf_vs_snprintf
!./sprintf_vs_snprintf
```

```python
%%writefile student_report_card_generator.c
#include <stdio.h>

int main()
{
    // ----------------------------------------------------
    // Mini Project: Student Report Card Generator
    //
    // Uses snprintf() to generate a professional
    // report card stored inside a string.
    // ----------------------------------------------------

    char report[1000];

    int roll = 101;
    char name[] = "Rahul Kumar";

    int c = 86;
    int cpp = 91;
    int python = 94;
    int java = 89;
    int dbms = 92;

    int total = c + cpp + python + java + dbms;
    float percentage = total / 5.0;

    char grade;

    if (percentage >= 90)
        grade = 'A';
    else if (percentage >= 80)
        grade = 'B';
    else if (percentage >= 70)
        grade = 'C';
    else if (percentage >= 60)
        grade = 'D';
    else
        grade = 'F';

    snprintf(report,
             sizeof(report),
             "=============== STUDENT REPORT CARD ===============\n"
             "Roll Number : %d\n"
             "Student Name: %s\n"
             "---------------------------------------------------\n"
             "C Programming      : %d\n"
             "C++ Programming    : %d\n"
             "Python Programming : %d\n"
             "Java Programming   : %d\n"
             "DBMS               : %d\n"
             "---------------------------------------------------\n"
             "Total Marks : %d / 500\n"
             "Percentage  : %.2f%%\n"
             "Grade       : %c\n"
             "Result      : %s\n"
             "===================================================\n",
             roll,
             name,
             c,
             cpp,
             python,
             java,
             dbms,
             total,
             percentage,
             grade,
             (grade == 'F') ? "FAIL" : "PASS");

    printf("%s", report);

    return 0;
}

!gcc student_report_card_generator.c -o student_report_card_generator
!./student_report_card_generator
```

```python
%%writefile binary_representation.c
#include <stdio.h>

void printBinary(int number)
{
    int i;

    for(i = 31; i >= 0; i--)
    {
        printf("%d", (number >> i) & 1);

        if(i % 4 == 0)
            printf(" ");
    }
}

int main()
{
    // Display binary representation

    int number = 25;

    printf("Decimal Number : %d\n", number);

    printf("Binary Number  : ");
    printBinary(number);

    printf("\n");

    return 0;
}

!gcc binary_representation.c -o binary_representation
!./binary_representation
```

```python
%%writefile sign_preserving_conversion.c
#include <stdio.h>

int main()
{
    // ----------------------------------------------------
    // Program: Sign-Preserving Conversion
    //
    // A sign-preserving conversion keeps the sign
    // of the original signed value when converting
    // to a wider signed type.
    // ----------------------------------------------------

    short smallValue = -250;

    int largeValue = smallValue;

    printf("Original short : %hd\n", smallValue);
    printf("Converted int  : %d\n", largeValue);

    return 0;
}

!gcc sign_preserving_conversion.c -o sign_preserving_conversion
!./sign_preserving_conversion
```

```python
%%writefile value_preserving_conversion.c
#include <stdio.h>

int main()
{
    // ----------------------------------------------------
    // Program: Value-Preserving Conversion
    //
    // Value-preserving conversion retains the
    // numeric value when converting to a type
    // that can represent it.
    // ----------------------------------------------------

    unsigned short value = 500;

    unsigned int convertedValue = value;

    printf("Original Value  : %hu\n", value);
    printf("Converted Value : %u\n", convertedValue);

    return 0;
}

!gcc value_preserving_conversion.c -o value_preserving_conversion
!./value_preserving_conversion
```

```python
%%writefile sign_vs_value_preserving_conversion.c
#include <stdio.h>

int main()
{
    // ----------------------------------------------------
    // Program: Sign-Preserving vs Value-Preserving
    // ----------------------------------------------------

    short signedValue = -100;
    unsigned short unsignedValue = 100;

    int signedResult = signedValue;
    unsigned int unsignedResult = unsignedValue;

    printf("Sign-Preserving Conversion\n");
    printf("--------------------------\n");
    printf("Original Value  : %hd\n", signedValue);
    printf("Converted Value : %d\n\n", signedResult);

    printf("Value-Preserving Conversion\n");
    printf("---------------------------\n");
    printf("Original Value  : %hu\n", unsignedValue);
    printf("Converted Value : %u\n", unsignedResult);

    return 0;
}

!gcc sign_vs_value_preserving_conversion.c -o sign_vs_value_preserving_conversion
!./sign_vs_value_preserving_conversion
```
