```python
%%writefile realloc_example.c
#include <stdio.h>
#include <stdlib.h>

int main()
{
    // ----------------------------------------------------
    // Program: realloc()
    // ----------------------------------------------------

    int *ptr;

    ptr = (int *)malloc(3 * sizeof(int));

    for(int i = 0; i < 3; i++)
    {
        ptr[i] = (i + 1) * 10;
    }

    ptr = (int *)realloc(ptr, 5 * sizeof(int));

    ptr[3] = 40;
    ptr[4] = 50;

    printf("Array after realloc():\n\n");

    for(int i = 0; i < 5; i++)
    {
        printf("%d ", ptr[i]);
    }

    printf("\n");

    free(ptr);

    return 0;
}
```

```python
%%writefile dynamic_memory_complete.c
#include <stdio.h>
#include <stdlib.h>

int main()
{
    // ----------------------------------------------------
    // Program: malloc(), realloc() and free()
    // ----------------------------------------------------

    int *ptr;

    // Allocate memory
    ptr = (int *)malloc(3 * sizeof(int));

    if(ptr == NULL)
    {
        printf("Memory Allocation Failed.\n");
        return 0;
    }

    for(int i = 0; i < 3; i++)
    {
        ptr[i] = (i + 1) * 100;
    }

    printf("After malloc():\n");

    for(int i = 0; i < 3; i++)
    {
        printf("%d ", ptr[i]);
    }

    printf("\n\n");

    // Reallocate memory
    ptr = (int *)realloc(ptr, 5 * sizeof(int));

    ptr[3] = 400;
    ptr[4] = 500;

    printf("After realloc():\n");

    for(int i = 0; i < 5; i++)
    {
        printf("%d ", ptr[i]);
    }

    printf("\n\n");

    // Release memory
    free(ptr);
    ptr = NULL;

    printf("Memory Released Successfully.\n");

    return 0;
}
```

```python
%%writefile allocate_array_of_pointers.c
#include <stdio.h>
#include <stdlib.h>

int main()
{
    // ----------------------------------------------------
    // Program: Allocating Memory for an Array of Pointers
    // ----------------------------------------------------

    int **ptr;

    ptr = (int **)malloc(3 * sizeof(int *));

    for(int i = 0; i < 3; i++)
    {
        ptr[i] = (int *)malloc(sizeof(int));

        *ptr[i] = (i + 1) * 100;
    }

    printf("Values:\n");

    for(int i = 0; i < 3; i++)
    {
        printf("%d ", *ptr[i]);
    }

    printf("\n");

    for(int i = 0; i < 3; i++)
    {
        free(ptr[i]);
    }

    free(ptr);

    return 0;
}
```

```python
%%writefile pointer_vs_array_pointers.c
#include <stdio.h>

int main()
{
    // ----------------------------------------------------
    // Program: Pointer to Array vs Array of Pointers
    // ----------------------------------------------------

    int numbers[3] = {10,20,30};

    int (*arrayPointer)[3] = &numbers;

    int *pointerArray[3];

    for(int i = 0; i < 3; i++)
    {
        pointerArray[i] = &numbers[i];
    }

    printf("Using Pointer to Array:\n");

    for(int i = 0; i < 3; i++)
    {
        printf("%d ", (*arrayPointer)[i]);
    }

    printf("\nUsing Array of Pointers:\n");

    for(int i = 0; i < 3; i++)
    {
        printf("%d ", *pointerArray[i]);
    }

    printf("\n");

    return 0;
}
```

```python
%%writefile dynamic_2d_array.c
#include <stdio.h>
#include <stdlib.h>

int main()
{
    // ----------------------------------------------------
    // Program: Dynamic 2D Array using Double Pointer
    // ----------------------------------------------------

    int rows = 3;
    int cols = 4;

    int **matrix;

    matrix = (int **)malloc(rows * sizeof(int *));

    for(int i = 0; i < rows; i++)
    {
        matrix[i] = (int *)malloc(cols * sizeof(int));
    }

    int value = 1;

    for(int i = 0; i < rows; i++)
    {
        for(int j = 0; j < cols; j++)
        {
            matrix[i][j] = value++;
        }
    }

    printf("Matrix:\n");

    for(int i = 0; i < rows; i++)
    {
        for(int j = 0; j < cols; j++)
        {
            printf("%3d ", matrix[i][j]);
        }

        printf("\n");
    }

    for(int i = 0; i < rows; i++)
    {
        free(matrix[i]);
    }

    free(matrix);

    return 0;
}
```

```python
%%writefile free_dynamic_2d.c
#include <stdio.h>
#include <stdlib.h>

int main()
{
    // ----------------------------------------------------
    // Program: Freeing Dynamic 2D Memory
    // ----------------------------------------------------

    int rows = 2;
    int cols = 3;

    int **matrix = (int **)malloc(rows * sizeof(int *));

    for(int i = 0; i < rows; i++)
    {
        matrix[i] = (int *)malloc(cols * sizeof(int));
    }

    printf("Memory Allocated Successfully\n");

    for(int i = 0; i < rows; i++)
    {
        free(matrix[i]);
    }

    free(matrix);

    printf("Memory Freed Successfully\n");

    return 0;
}
```

```python
%%writefile static_vs_dynamic_2d.c
#include <stdio.h>
#include <stdlib.h>

int main()
{
    // ----------------------------------------------------
    // Program: Static vs Dynamic 2D Arrays
    // ----------------------------------------------------

    int staticMatrix[2][2] =
    {
        {1, 2},
        {3, 4}
    };

    int **dynamicMatrix = (int **)malloc(2 * sizeof(int *));

    for(int i = 0; i < 2; i++)
    {
        dynamicMatrix[i] = (int *)malloc(2 * sizeof(int));
    }

    dynamicMatrix[0][0] = 10;
    dynamicMatrix[0][1] = 20;
    dynamicMatrix[1][0] = 30;
    dynamicMatrix[1][1] = 40;

    printf("Static Matrix:\n");

    for(int i = 0; i < 2; i++)
    {
        for(int j = 0; j < 2; j++)
        {
            printf("%3d ", staticMatrix[i][j]);
        }

        printf("\n");
    }

    printf("\nDynamic Matrix:\n");

    for(int i = 0; i < 2; i++)
    {
        for(int j = 0; j < 2; j++)
        {
            printf("%3d ", dynamicMatrix[i][j]);
        }

        printf("\n");
    }

    for(int i = 0; i < 2; i++)
    {
        free(dynamicMatrix[i]);
    }

    free(dynamicMatrix);

    return 0;
}
```

```python
%%writefile dynamic_matrix_project.c
#include <stdio.h>
#include <stdlib.h>

int main()
{
    // ----------------------------------------------------
    // Mini Project:
    // Dynamic Matrix Operations using Double Pointers
    // ----------------------------------------------------

    int rows = 3;
    int cols = 3;

    int **matrix = (int **)malloc(rows * sizeof(int *));

    for(int i = 0; i < rows; i++)
    {
        matrix[i] = (int *)malloc(cols * sizeof(int));
    }

    int value = 1;

    printf("Generated Matrix:\n");

    for(int i = 0; i < rows; i++)
    {
        for(int j = 0; j < cols; j++)
        {
            matrix[i][j] = value++;
            printf("%3d ", matrix[i][j]);
        }

        printf("\n");
    }

    int sum = 0;

    for(int i = 0; i < rows; i++)
    {
        for(int j = 0; j < cols; j++)
        {
            sum += matrix[i][j];
        }
    }

    printf("\nSum of Matrix Elements = %d\n", sum);

    for(int i = 0; i < rows; i++)
    {
        free(matrix[i]);
    }

    free(matrix);

    return 0;
}
```
