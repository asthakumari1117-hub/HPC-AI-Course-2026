```python
%%writefile dynamic_memory_allocation.c
#include <stdio.h>
#include <stdlib.h>

int main()
{
    // ----------------------------------------------------
    // Allocate memory dynamically using malloc().
    // ----------------------------------------------------

    int *ptr;

    ptr = (int *)malloc(sizeof(int));

    if(ptr == NULL)
    {
        printf("Memory Allocation Failed!\n");
        return 1;
    }

    *ptr = 500;

    printf("Stored Value : %d\n", *ptr);
    printf("Memory Address : %p\n", ptr);

    free(ptr);

    return 0;
}
```

```python
%%writefile calloc_demo.c
#include <stdio.h>
#include <stdlib.h>

int main()
{
    // ----------------------------------------------------
    // Program: Dynamic Memory Allocation using calloc()
    // calloc() allocates memory and initializes all bytes
    // to zero.
    // ----------------------------------------------------

    int *ptr;

    ptr = (int *)calloc(5, sizeof(int));

    if(ptr == NULL)
    {
        printf("Memory Allocation Failed!\n");
        return 1;
    }

    printf("Elements after calloc():\n");

    for(int i = 0; i < 5; i++)
    {
        printf("%d ", ptr[i]);
    }

    free(ptr);

    return 0;
}
```

```python
%%writefile search_node.c
#include <stdio.h>
#include <stdlib.h>

struct Node
{
    int data;
    struct Node *next;
};

int main()
{
    // ----------------------------------------------------
    // Program: Search an Element
    // ----------------------------------------------------

    struct Node *first, *second, *third, *temp;

    int key = 20;
    int found = 0;

    first = (struct Node *)malloc(sizeof(struct Node));
    second = (struct Node *)malloc(sizeof(struct Node));
    third = (struct Node *)malloc(sizeof(struct Node));

    first->data = 10;
    second->data = 20;
    third->data = 30;

    first->next = second;
    second->next = third;
    third->next = NULL;

    temp = first;

    while(temp != NULL)
    {
        if(temp->data == key)
        {
            found = 1;
            break;
        }

        temp = temp->next;
    }

    if(found)
        printf("%d Found\n", key);
    else
        printf("%d Not Found\n", key);

    free(first);
    free(second);
    free(third);

    return 0;
}
```

```python
%%writefile count_circular_nodes.c
#include <stdio.h>
#include <stdlib.h>

struct Node
{
    int data;
    struct Node *next;
};

int main()
{
    // ----------------------------------------------------
    // Program: Count Nodes in Circular Linked List
    // ----------------------------------------------------

    struct Node *first, *second, *third, *temp;

    int count = 0;

    first = (struct Node *)malloc(sizeof(struct Node));
    second = (struct Node *)malloc(sizeof(struct Node));
    third = (struct Node *)malloc(sizeof(struct Node));

    first->next = second;
    second->next = third;
    third->next = first;

    temp = first;

    do
    {
        count++;
        temp = temp->next;
    }
    while(temp != first);

    printf("Total Nodes = %d\n", count);

    free(first);
    free(second);
    free(third);

    return 0;
}
```

```python
%%writefile reverse_linked_list.c
#include <stdio.h>
#include <stdlib.h>

struct Node
{
    int data;
    struct Node *next;
};

int main()
{
    // ----------------------------------------------------
    // Program: Reverse a Singly Linked List
    // ----------------------------------------------------

    struct Node *first,*second,*third;
    struct Node *prev,*current,*next;

    first=(struct Node*)malloc(sizeof(struct Node));
    second=(struct Node*)malloc(sizeof(struct Node));
    third=(struct Node*)malloc(sizeof(struct Node));

    first->data=10;
    second->data=20;
    third->data=30;

    first->next=second;
    second->next=third;
    third->next=NULL;

    prev=NULL;
    current=first;

    while(current!=NULL)
    {
        next=current->next;
        current->next=prev;
        prev=current;
        current=next;
    }

    first=prev;

    printf("Reversed Linked List:\n");

    current=first;

    while(current!=NULL)
    {
        printf("%d -> ",current->data);
        current=current->next;
    }

    printf("NULL\n");

    free(third);
    free(second);
    free(first);

    return 0;
}
```

```python
%%writefile sort_linked_list.c
#include <stdio.h>
#include <stdlib.h>

struct Node
{
    int data;
    struct Node *next;
};

int main()
{
    // ----------------------------------------------------
    // Program: Sort a Singly Linked List
    // ----------------------------------------------------

    struct Node *first,*second,*third;
    struct Node *current,*index;

    int temp;

    first=(struct Node*)malloc(sizeof(struct Node));
    second=(struct Node*)malloc(sizeof(struct Node));
    third=(struct Node*)malloc(sizeof(struct Node));

    first->data=30;
    second->data=10;
    third->data=20;

    first->next=second;
    second->next=third;
    third->next=NULL;

    for(current=first;current!=NULL;current=current->next)
    {
        for(index=current->next;index!=NULL;index=index->next)
        {
            if(current->data>index->data)
            {
                temp=current->data;
                current->data=index->data;
                index->data=temp;
            }
        }
    }

    printf("Sorted Linked List:\n");

    current=first;

    while(current!=NULL)
    {
        printf("%d -> ",current->data);
        current=current->next;
    }

    printf("NULL\n");

    free(first);
    free(second);
    free(third);

    return 0;
}
```

```python
%%writefile complete_stack_linked_list.c
#include <stdio.h>
#include <stdlib.h>

struct Node
{
    int data;
    struct Node *next;
};

int main()
{
    // ----------------------------------------------------
    // Program: Push, Peek, Pop and Display
    // ----------------------------------------------------

    struct Node *top = NULL;
    struct Node *newNode;
    struct Node *temp;

    // Push
    for(int i = 10; i <= 30; i += 10)
    {
        newNode = (struct Node *)malloc(sizeof(struct Node));

        newNode->data = i;
        newNode->next = top;
        top = newNode;
    }

    // Peek
    printf("Top Element = %d\n\n", top->data);

    // Pop
    printf("Deleted Element = %d\n\n", top->data);

    temp = top;
    top = top->next;
    free(temp);

    // Display
    printf("Stack After Pop:\n\n");

    temp = top;

    while(temp != NULL)
    {
        printf("| %d |\n", temp->data);
        temp = temp->next;
    }

    while(top != NULL)
    {
        temp = top;
        top = top->next;
        free(temp);
    }

    return 0;
}
```

```python
%%writefile enqueue_operation.c
#include <stdio.h>

#define SIZE 5

int main()
{
    // ----------------------------------------------------
    // Program: Enqueue Operation
    // ----------------------------------------------------

    int queue[SIZE];
    int front = -1;
    int rear = -1;

    if(front == -1)
        front = 0;

    queue[++rear] = 10;
    queue[++rear] = 20;
    queue[++rear] = 30;

    printf("Queue After Enqueue:\n\n");

    for(int i = front; i <= rear; i++)
    {
        printf("%d ", queue[i]);
    }

    return 0;
}
```

```python
%%writefile dequeue_operation.c
#include <stdio.h>

#define SIZE 5

int main()
{
    // ----------------------------------------------------
    // Program: Dequeue Operation
    // ----------------------------------------------------

    int queue[SIZE] = {10,20,30};

    int front = 0;
    int rear = 2;

    printf("Deleted Element = %d\n\n", queue[front]);

    front++;

    printf("Queue After Dequeue:\n\n");

    for(int i = front; i <= rear; i++)
    {
        printf("%d ", queue[i]);
    }

    return 0;
}
```

```python
%%writefile complete_queue_linked_list.c
#include <stdio.h>
#include <stdlib.h>

struct Node
{
    int data;
    struct Node *next;
};

int main()
{
    // ----------------------------------------------------
    // Program: Enqueue, Front, Rear, Dequeue & Display
    // ----------------------------------------------------

    struct Node *front = NULL;
    struct Node *rear = NULL;
    struct Node *newNode;
    struct Node *temp;

    // Enqueue
    for(int i = 10; i <= 30; i += 10)
    {
        newNode = (struct Node *)malloc(sizeof(struct Node));

        newNode->data = i;
        newNode->next = NULL;

        if(front == NULL)
            front = rear = newNode;
        else
        {
            rear->next = newNode;
            rear = newNode;
        }
    }

    printf("Front Element = %d\n", front->data);
    printf("Rear Element  = %d\n\n", rear->data);

    printf("Deleted Element = %d\n\n", front->data);

    temp = front;
    front = front->next;
    free(temp);

    if(front == NULL)
        rear = NULL;

    printf("Queue After Dequeue:\n\n");

    temp = front;

    while(temp != NULL)
    {
        printf("%d ", temp->data);
        temp = temp->next;
    }

    printf("\n");

    return 0;
}
```
