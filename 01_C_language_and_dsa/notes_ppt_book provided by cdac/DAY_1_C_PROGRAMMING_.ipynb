```python
%%writefile ieee754_intro.c
#include <stdio.h>

int main()
{
    // Computers store floating point numbers
    // using the IEEE-754 Standard.
    //
    // A floating point number consists of:
    //
    // -----------------------------------------
    // | Sign | Exponent | Fraction (Mantissa) |
    // -----------------------------------------
    //
    // float (32-bit)
    //
    // -----------------------------------------
    // |1 Bit| 8 Bits |       23 Bits          |
    // -----------------------------------------
    //
    // double (64-bit)
    //
    // -----------------------------------------
    // |1 Bit|11 Bits |       52 Bits          |
    // -----------------------------------------
    //
    // Sign     -> Positive or Negative
    // Exponent -> Magnitude of the number
    // Mantissa -> Precision of the number
    //
    // IEEE-754 allows computers to represent
    // decimal values efficiently.

    float number = 3.1415926f;

    printf("Floating Point Number : %.7f\n", number);

    return 0;
}

!gcc ieee754_intro.c -o ieee754_intro
!./ieee754_intro
```

```python
%%writefile data_alignment.c
#include <stdio.h>

int main()
{
    // Data Alignment means storing
    // variables at memory addresses
    // suitable for the processor.
    //
    // Proper alignment allows the CPU
    // to access data more efficiently.
    //
    // Example:
    //
    // int (4 Bytes)
    //
    // Good Alignment
    // Address : 100,104,108,...
    //
    // Poor Alignment
    // Address : 101,105,...
    //
    // Modern compilers automatically
    // align variables.

    int number = 100;

    printf("Value   : %d\n", number);
    printf("Address : %p\n", (void*)&number);

    return 0;
}

!gcc data_alignment.c -o data_alignment
!./data_alignment
```

```python
%%writefile memset_intro.c
#include <stdio.h>
#include <string.h>

int main()
{
    // memset()
    // initializes memory
    // with a specific byte.
    //
    // Syntax:
    // memset(pointer, value, bytes);

    char text[20];

    memset(text, '*', sizeof(text));

    for(int i = 0; i < sizeof(text); i++)
        printf("%c ", text[i]);

    printf("\n");

    return 0;
}

!gcc memset_intro.c -o memset_intro
!./memset_intro
```

```python
%%writefile memcmp_intro.c
#include <stdio.h>
#include <string.h>

int main()
{
    // memcmp()
    // compares two
    // memory blocks.
    //
    // Return Value:
    //
    // < 0
    // = 0
    // > 0

    char first[] = "APPLE";
    char second[] = "APPLE";

    int result = memcmp(first, second, sizeof(first));

    printf("Comparison Result = %d\n", result);

    if(result == 0)
        printf("Memory Blocks are Equal\n");
    else
        printf("Memory Blocks are Different\n");

    return 0;
}

!gcc memcmp_intro.c -o memcmp_intro
!./memcmp_intro
```

```python
%%writefile matrix_addition.c
#include <stdio.h>

int main()
{
    // Matrix Addition
    // Both matrices must
    // have the same order.

    int A[2][2] = {{1,2},{3,4}};
    int B[2][2] = {{5,6},{7,8}};
    int C[2][2];

    for(int i=0;i<2;i++)
    {
        for(int j=0;j<2;j++)
        {
            C[i][j]=A[i][j]+B[i][j];
        }
    }

    printf("Result Matrix\n");

    for(int i=0;i<2;i++)
    {
        for(int j=0;j<2;j++)
            printf("%4d",C[i][j]);

        printf("\n");
    }

    return 0;
}

!gcc matrix_addition.c -o matrix_addition
!./matrix_addition
```

```python
%%writefile matrix_multiplication.c
#include <stdio.h>

int main()
{
    // Matrix multiplication
    // Rule:
    // Columns of first matrix
    // must equal rows of second.

    int A[2][2]={{1,2},{3,4}};
    int B[2][2]={{5,6},{7,8}};
    int C[2][2]={0};

    for(int i=0;i<2;i++)
    {
        for(int j=0;j<2;j++)
        {
            for(int k=0;k<2;k++)
            {
                C[i][j]+=A[i][k]*B[k][j];
            }
        }
    }

    printf("Multiplication Result\n");

    for(int i=0;i<2;i++)
    {
        for(int j=0;j<2;j++)
            printf("%5d",C[i][j]);

        printf("\n");
    }

    return 0;
}

!gcc matrix_multiplication.c -o matrix_multiplication
!./matrix_multiplication
```

```python
%%writefile symmetric_matrix.c
#include <stdio.h>

int main()
{
    // Symmetric Matrix
    // A[i][j] = A[j][i]

    int matrix[3][3]=
    {
        {1,2,3},
        {2,4,5},
        {3,5,6}
    };

    int symmetric=1;

    for(int i=0;i<3;i++)
    {
        for(int j=0;j<3;j++)
        {
            if(matrix[i][j]!=matrix[j][i])
                symmetric=0;
        }
    }

    if(symmetric)
        printf("Matrix is Symmetric\n");
    else
        printf("Matrix is Not Symmetric\n");

    return 0;
}

!gcc symmetric_matrix.c -o symmetric_matrix
!./symmetric_matrix
```

```python
%%writefile matrix_calculator.c
#include <stdio.h>

int main()
{
    // Mini Project
    // Matrix Calculator
    // Performs Addition,
    // Subtraction and
    // Multiplication.

    int A[2][2]={{2,4},{6,8}};
    int B[2][2]={{1,3},{5,7}};
    int add[2][2];
    int sub[2][2];
    int mul[2][2]={0};

    for(int i=0;i<2;i++)
    {
        for(int j=0;j<2;j++)
        {
            add[i][j]=A[i][j]+B[i][j];
            sub[i][j]=A[i][j]-B[i][j];

            for(int k=0;k<2;k++)
                mul[i][j]+=A[i][k]*B[k][j];
        }
    }

    printf("Addition\n");
    for(int i=0;i<2;i++)
    {
        for(int j=0;j<2;j++)
            printf("%4d",add[i][j]);
        printf("\n");
    }

    printf("\nSubtraction\n");
    for(int i=0;i<2;i++)
    {
        for(int j=0;j<2;j++)
            printf("%4d",sub[i][j]);
        printf("\n");
    }

    printf("\nMultiplication\n");
    for(int i=0;i<2;i++)
    {
        for(int j=0;j<2;j++)
            printf("%5d",mul[i][j]);
        printf("\n");
    }

    return 0;
}

!gcc matrix_calculator.c -o matrix_calculator
!./matrix_calculator
```

```python
%%writefile student_result_management.c
#include <stdio.h>

int main()
{
    // Mini Project
    //
    // Store marks of
    // multiple students
    // in multiple subjects
    // using a 2D array.

    int marks[3][4] =
    {
        {85,90,88,92},
        {78,81,75,80},
        {91,89,94,90}
    };

    printf("============================================\n");
    printf("     STUDENT RESULT MANAGEMENT SYSTEM\n");
    printf("============================================\n");

    for(int i=0;i<3;i++)
    {
        int total=0;

        printf("\nStudent %d\n",i+1);

        for(int j=0;j<4;j++)
        {
            printf("Subject %d : %d\n",j+1,marks[i][j]);
            total+=marks[i][j];
        }

        printf("Total   : %d\n",total);
        printf("Average : %.2f\n",(float)total/4);
    }

    return 0;
}

!gcc student_result_management.c -o student_result_management
!./student_result_management
```
